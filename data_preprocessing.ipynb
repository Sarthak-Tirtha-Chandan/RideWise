## Importing Modules

```python
import pandas as pd 
import numpy as np
from matplotlib import pyplot as plt 
import seaborn as sns
```

## Loading the Dataset

```python
df_hour = pd.read_csv('data/raw/hour.csv')

df_hour_clean = df_hour.copy()
df_hour_clean.head()
```

## Data Cleaning


```python
df_hour_clean.info()
```

### Convert dteday to DateTime

```python
df_hour_clean["dteday"] = pd.to_datetime(df_hour_clean["dteday"])

df_hour_clean.info()
```

### Create Datetime column 
datetime = dteday + hr

```python
df_hour_clean['datetime'] = (
    df_hour_clean['dteday'] +
    pd.to_timedelta(df_hour_clean['hr'] , unit='hr')
)

df_hour_clean.head()
```

```python
# Drop dteday

df_hour_clean = df_hour_clean.drop(columns=['dteday'] , axis=1)
df_hour_clean.head()
```

### Sort Chronologically
Time series models require data in time order

```python
df_hour_clean = df_hour_clean.sort_values('datetime')
df_hour_clean.head(7)
```

```python
df_hour_clean = df_hour_clean.reset_index(drop=True)
df_hour_clean.head(7)
```

### Creating datetime as index

```python
df_hour_clean.set_index('datetime' , inplace=True)
df_hour_clean.head()
```

### Removing unnessasary columns

```python
df_hour_clean.drop(columns=['instant'] , axis=1 , inplace=True)
df_hour_clean.head()
```

### Removing leakage columns

The dataset defines:

cnt = casual + registered

If the model receives both casual and registered, it can simply compute:
100 + 250 = 350

It doesn't need to learn anything.

```python
df_hour_clean.drop(columns=['casual' , 'registered'] , axis=1 , inplace=True)
df_hour_clean.head()
```

### Saving the Data

```python
df_hour_clean.to_csv('data/processed/hour_clean.csv')
```

# Feature Engineering

```python
df = df_hour_clean.copy()
df.head()
```

### Cyclic Encoding

Time is cyclic, not linear.<br>
Linear numbers make 23 and 0 look far apart, even though they're consecutive.<br>
So we form a cycle using sine and cosine values

```python
# Hour 
df["hr_sin"] = np.sin(2 * np.pi * df["hr"] / 24)
df["hr_cos"] = np.cos(2 * np.pi * df["hr"] / 24)

# Week
df["week_sin"] = np.sin(2 * np.pi * df["weekday"] / 7)
df["week_cos"] = np.cos(2 * np.pi * df["weekday"] / 7)

# Month
df["mnth_sin"] = np.sin(2 * np.pi * df["mnth"] / 12)
df["mnth_cos"] = np.cos(2 * np.pi * df["mnth"] / 12)

df.head()
```

### Creating Lag Features

Creating<br>
lag_1 : previous hour demand <br>
lag_24 : previous day demand in same hour <br>
lag_168 : previous week demand in same hour


```python
df['lag_1'] = df['cnt'].shift(1)
df['lag_24'] = df['cnt'].shift(24)
df['lag_168'] = df['cnt'].shift(168)

df.head()
```

### Rolling Window Features

Instead of looking at one previous value,
look at the average of several previous values
<br>
rolling mean , median , standard deviation

```python
df['rolling_mean'] = df['cnt'].rolling(window=24).mean()
df['rolling_median'] = df['cnt'].rolling(window=24).median()
df['rolling_std'] = df['cnt'].rolling(window=24).std()

df.tail()
```

### Adding peak hours column

```python
df["is_peak_hour"] = (
    df["hr"].isin([7,8,9,17,18,19])
).astype(int)

df.sample(5)
```

### Adding Interaction Features

weather x hr

temp x hr 

temp x humidity

```python
df['weather_hr'] = df['weathersit'] * df['hr']
df['temp_hr'] = df['temp'] * df['hr']
df['temp_hum'] = df['temp'] * df['hum']

df.head()
```

### Handling Missing values

```python
df.isna().sum()
```

```python
df.dropna(inplace=True)
df.isna().sum()
```

#### Saving df

```python
df.to_csv('data/processed/hour_features.csv')
```
